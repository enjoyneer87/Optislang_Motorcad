## star of slots Theory

기계적으로 서로 인접한 슬롯이지만 전기적으로는 위상차를 가져야된다.
S는 슬롯의 개수, p는 극상수이다.
따라서 \theta는 Star of Slot은 전기적 위상차를 가져야 한다.


$ \theta =  \frac{2\pi p}{S}$

$\alpha_p$는 각 위상이 층수에 따라 가질수 있는 화살표의 수이다. 

$\alpha_p=\frac{S l}{m}$

m은 위상수이다. 3상의 경우 m=3이다.

number of spokes in the star of slots

```python
poleNumber=10
slotNumber=12
phaseNumber=3
def getNumberofSpkes(poleNumber,slotNumber):
    periodicity=np.gcd(poleNumber,slotNumber)
    numberofspke=slotNumber/periodicity
    return numberofspke
```

angle between two spokes

```python
def getAngleBetweenSpkes(poleNumber,slotNumber):
    periodicity=np.gcd(poleNumber,slotNumber)
    a_ph=360*periodicity/slotNumber # the angle between two spokes
    return a_ph
a_ph=getAngleBetweenSpkes(poleNumber,slotNumber)
print("a_ph=",a_ph)
```

the star of slots divided into 2*m sector

```python
def getSpanofStar(phaseNumber):
    spanofStar=360/(2*phaseNumber)
    return spanofStar

spanofStar=getSpanofStar(phaseNumber)
spanofStar
```

the number of spokes in each sector

```python
def getNumberSpokeInEachSector(poleNumber,slotNumber,phaseNumber):
    numberofspke=getNumberofSpkes(poleNumber,slotNumber)
    NumberSpokeInEachSector=numberofspke/(2*phaseNumber)
    # == a_ph/spanofStar
    return NumberSpokeInEachSector
```

MMF 계산 (Winding Function)

\begin{aligned}
& f(\theta)=a_0+\sum_{n=1}^{\infty}\left(a_n \cos n \theta+b_n \sin n \theta\right) 
&\quad=\sum_{n=1}^{\infty}\left(a_n \cos n \theta\right) \quad\left(\because a_0=b_n=0\right)
\end{aligned}

```python
import numpy as np

mu_0 = 4 * np.pi * 1e-7  # Permeability of free space

def B_arm_slotless(theta_vect, t_vect, machine_params):
    # Load machine parameters
    p = machine_params['p']                               # Number of pole pairs
    s = machine_params['s']                               # Number of slots
    theta_o = machine_params['theta_o']                   # Slot opening pitch
    N_t = machine_params['n_s'] * machine_params['N_c'] / machine_params['N_ph']  # Turns per phase
    R_s = machine_params['R_s']                           # Inner stator radius
    R_r = machine_params['R_r']                           # Rotor radius
    I_1 = machine_params['i_RMS'] * np.sqrt(2)            # Amplitude of fundamental current
    omega_m = machine_params['omega_m']                   # Rotation speed

    # Spatial orders, odd, non-triplen multiples of F1
    F_1 = np.gcd(p, s)                                    # Number of parts
    v_seg = np.arange(1, 51, 2)
    v_seg = v_seg[v_seg % 3 != 0]

    v = np.transpose(v_seg) * F_1

    # Radial and tangential magnitudes
    B_mv_r = 3 * mu_0 / 2 * J_v(N_t, R_s, v, s, theta_o) * F_v(R_r, R_s, v) * I_1
    B_mv_t = 3 * mu_0 / 2 * J_v(N_t, R_s, v, s, theta_o) * G_v(R_r, R_s, v) * I_1

    # Create theta-t matrix and loop through time
    B_arm = np.zeros((len(t_vect), len(theta_vect)), dtype=complex)

    for i in range(len(t_vect)):
        B_arm[i, :] = np.sum(B_mv_r * np.sin(rot(v, F_1) * v * theta_vect - p * omega_m * t_vect[i]) +
                             1j * B_mv_t * rot(v, F_1) * np.cos(rot(v, F_1) * v * theta_vect - p * omega_m * t_vect[i]), axis=0)

    return B_arm

def J_v(N_t, R_s, v, s, theta_o):
    # J_m according to Zhu (2010)
    return 2 / np.pi * N_t / R_s * K_sov(v, theta_o) * K_pv(v, s) * K_dv(v)

def K_sov(v, theta_o):
    # The slot opening factor from Zhu (2010)
    return np.sin(v * theta_o / 2) / (v * theta_o / 2)

def K_pv(v, s):
    # The pitch factor as described by Zhu and Gieras
    return np.sin(v * np.pi / s)

def K_dv(v):
    # The distribution factor is equal to 1
    return np.ones_like(v)

def rot(v, F1):
    # The rotation factor according to Yokoi (2016)
    k_taum = np.round(np.sin(np.pi * (v / F1 - 1) / 2), 5) / np.round((3 * np.sin(np.pi * (v / F1 - 1) / 6)), 5)
    k_taum[np.isnan(k_taum)] = 1
    k_taup = np.round(np.sin(np.pi * (v / F1 + 1) / 2), 5) / np.round((3 * np.sin(np.pi * (v / F1 + 1) / 6)), 5)
    k_taup[np.isnan(k_taup)] = 1

    return np.sign(k_taup - k_taum)

def F_v(R_r, R_s, v):
    # The geometrical function of Zhu (2010) for r = R_s
    return (1 + (R_r / R_s) ** (2 * v)) / (1 - (R_r / R_s) ** (2 * v))

def G_v(R_r, R_s, v):
    # Tangential geometric function of Zhu (2010) equals 1 for r = R_s
    return np.ones_like(v)
```
